## Descripción

Este programa calcula el caudal total, el consumo diario y el consumo mensual de agua a partir del número de plumas y su caudal individual.

```python
"""
Actividad Integradora: La Pluma de Agua

Autor: Juliana Mercado

Este programa calcula el consumo de agua
a partir del número de plumas instaladas.
"""

# Solicitar datos al usuario
numero_plumas = int(input("Ingrese el número de plumas: "))
caudal_pluma = float(input("Ingrese el caudal de una pluma (L/s): "))

# Constante
SEGUNDOS_DIA = 86400

# Cálculo del caudal total
caudal_total = numero_plumas * caudal_pluma

# Consumo diario en litros
consumo_diario = caudal_total * SEGUNDOS_DIA

# Consumo mensual en litros
consumo_mensual = consumo_diario * 30

# Clasificación del consumo
if consumo_diario < 1000:
    nivel = "Bajo"
elif consumo_diario <= 3000:
    nivel = "Medio"
else:
    nivel = "Alto"

# Resultados
print("\nRESULTADOS")
print("--------------------")
print("Caudal total:", caudal_total, "L/s")
print("Consumo diario:", consumo_diario, "litros")
print("Consumo mensual:", consumo_mensual, "litros")
print("Nivel:", nivel)
```
