## Introducción

La expresión "pluma de agua" tiene un origen histórico relacionado con los sistemas hidráulicos utilizados durante la época colonial. En esta actividad se emplea dicho concepto para desarrollar un programa en Python que permita calcular el caudal y el consumo de agua de una vivienda.

A través de este ejercicio se integran conocimientos de historia, cultura, ingeniería y programación.

## Algoritmo en lenguaje natural

1. Solicitar al usuario el número de plumas de agua.
2. Definir el valor de una pluma en litros por segundo.
3. Definir la cantidad de segundos que tiene un día.
4. Calcular el caudal total.
5. Calcular el consumo diario.
6. Calcular el consumo mensual.
7. Clasificar el consumo como bajo, medio o alto.
8. Mostrar los resultados.

## Seudocódigo

Inicio

Leer número de plumas

Definir caudal de una pluma

Definir segundos por día

Calcular caudal total

Calcular consumo diario

Calcular consumo mensual

Si consumo diario < 1000

    Nivel = Bajo

Si no, si consumo diario ≤ 3000

    Nivel = Medio

En otro caso

    Nivel = Alto

Mostrar resultados

Fin

```python
# ==========================================================
# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON
# ==========================================================

# Solicitar al usuario el número de plumas de agua
plumas = float(input("Ingrese el número de plumas de agua: "))

# ----------------------------------------------------------
# Definir constantes del problema
# ----------------------------------------------------------

# Una pluma equivale a 0.025 litros por segundo
CAUDAL_PLUMA = 0.025

# Cantidad de segundos en un día
SEGUNDOS_DIA = 24 * 60 * 60

# ----------------------------------------------------------
# Calcular el caudal total
# ----------------------------------------------------------

caudal_total = plumas * CAUDAL_PLUMA

# ----------------------------------------------------------
# Calcular el consumo diario
# ----------------------------------------------------------

consumo_diario = caudal_total * SEGUNDOS_DIA

# ----------------------------------------------------------
# Calcular el consumo mensual
# ----------------------------------------------------------

consumo_mensual = consumo_diario * 30

# ----------------------------------------------------------
# Clasificar el consumo
# ----------------------------------------------------------

if consumo_diario < 1000:
    nivel = "Bajo"
elif consumo_diario <= 3000:
    nivel = "Medio"
else:
    nivel = "Alto"

# ----------------------------------------------------------
# Mostrar resultados
# ----------------------------------------------------------

print("\n========== RESULTADOS ==========")
print(f"Número de plumas: {plumas:.0f}")
print(f"Caudal total: {caudal_total:.3f} litros/segundo")
print(f"Consumo diario: {consumo_diario:.2f} litros")
print(f"Consumo mensual: {consumo_mensual:.2f} litros")
print(f"Clasificación del consumo: {nivel}")
```
